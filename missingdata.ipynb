missing data :

```python
import numpy as np
import pandas as pd
```

np.nan--> null value

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
```

```python
df=pd.DataFrame(data)
df
```

df.isna() --> gives value as true for a null value

```python
df.isna()
```

df.isna().sum() --> gives total no. of null values in each col

```python
df.isna().sum()
```

df.isna().any() --> if theres is any missing value in each column

```python
df.isna().any()
```

```python
df
```

removing missing data --> works on the basis of rows

df.dropna() --> drops the rows containing null values

```python
df.dropna()
```

df.dropna(thresh=3) --> threshold set to 3(the rows must contain atleast thresh non null values )

```python
df.dropna(thresh=1)
```

filling missing data

```python
df.fillna(0)
```

```python
values={'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

df.fillna(df.mean()) --> fills null value with mean of all values

```python
df.fillna(df.mean())
```
